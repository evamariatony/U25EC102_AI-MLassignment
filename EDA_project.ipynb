6. Identify numerical and categorical features

```python
new_df['bhk'] =new_df['size'].str.extract(r'(\d+)').astype(float)
```

```python
def handle_sqft_ranges(value):
    parts = str(value).split('-')
    if len(parts) == 2:
        return (float(parts[0]) + float(parts[1])) / 2
    try:
        return float(value)
    except:
        return None
df['total_sqft'] = df['total_sqft'].apply(handle_sqft_ranges)
```

8. Create at least 2 countplots/bar plots

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(4, 5))
sns.set_theme(style="whitegrid")
sns.countplot(data=new_df, x='area_type',hue='area_type',legend=False, palette='pastel', edgecolor='black')
plt.title("Distribution of Apartment Area Types", fontsize=12, fontweight='bold')
plt.xlabel("Area Type",fontsize=10)
plt.ylabel("Number of houses")
plt.show()
```

Through this count plot it can be interpreted that super built up area has had around 6000 listings while, built up areas rank second with a little over 1000 listings.In contrast it can be observed that Plot areas and Carpet areas are less and lesser utilized spaces respectively.

```python
plt.figure(figsize=(4,5))
std_bhk=new_df[new_df['bhk'].between(1,5)]
sns.countplot(std_bhk,x='bhk',hue='bhk', legend=False, palette='Blues_d', edgecolor='black')
plt.title("Most Common Apartment Configurations (BHK)", fontsize=13, fontweight='bold')
plt.xlabel("Number of Bedrooms (BHK)")
plt.ylabel("Number of Available Properties")
```

It can be clearly observed from the count plot that,2BHK properties are the market favorite and make up almost 3400 of the listings while,3BHK appartments are a close second making up around 3100 of those listings.

9. Create at least 1 boxplot.

```python
import matplotlib.pyplot as plt
import seaborn as sns
popular_places = ['Whitefield', 'Electronic City', 'Sarjapur Road', 'Kanakpura Road', 'Thanisandra']
df_filtered = new_df[new_df['location'].isin(popular_places)]
plt.figure(figsize=(10, 5))


sns.boxplot(data=df_filtered, x='location', y='price',hue='location', palette='Set2')
plt.ylim(0, 300)
plt.title("Price Distribution Across Top Locations")
plt.show()
```

Through this box plot it can be observed that white field has the overall highest pricing spectrum with a median over all othe locations at around 75lakhs and with a wide interquartile range with outliners well over the range,stretching to 3 Crore.Conversely Electronic city is seen to have the least median price with a highly compressed box,signifying highly standardized,uniform and predictable pricings.

10. Create at least 1 scatter plot.

```python
popular_places = ['Whitefield', 'Electronic City', 'Sarjapur Road', 'Kanakpura Road', 'Thanisandra']
df_filtered = new_df[new_df['location'].isin(popular_places)]
df_filtered['total_sqft'] = pd.to_numeric(df_filtered['total_sqft'], errors='coerce')
sns.scatterplot(data=df_filtered,x="total_sqft",y="price",hue="location",style="location",palette="deep")
plt.xlim(0,6000)
plt.ylim(0,1200)
plt.xlabel("total square feet")
plt.ylabel("price")
plt.show()
```

This plot illustrates a proportionaly increase in price with square feet.The location has also been made known and there is an observed regional clustering.Kanakpura road and Electronic city makes up locations in the under 2000 sqft and under 2 crore bracket,while Whitefield represents the upper tier oh housing well ranging from a 2000 to 6000 sqft housing and 3-8+ crore pricing scale.

```python
sns.scatterplot(data=df_filtered,x="total_sqft",y="bhk",hue="bhk",palette="deep",size="bhk",sizes=(40,200),marker='o',alpha=0.7)
plt.xlabel("total square feet")
plt.ylabel("bhk")
plt.xlim(0,4000)
plt.show()
```

This scatter plot between each bhk tier and total square feet show that each tier occupies a certain sqft range, which happens to increase in sixe as well as move towards the left side of the graph showing an increase in square feet with each tier.

11. Create a correlation heatmap.

```python
correlation=new_df.corr(numeric_only=True)
correlation
```

```python
sns.heatmap(correlation,cmap="RdBu_r",vmin=-1,vmax=1,fmt=".2f",linewidth=1,linecolor="black",annot=True)
```

This heat map illustrates the correlation among different aspects of housing,through colour gradient.It can be observed that bath and bhk have a strong correlation(0.82),showing that bedroom and bathroom count almost match.Price to shows strong correlation with bhk and bath.On the other hand balcony exhibits the weakest positive relation with all other variables.

12. Perform one additional visualization of your choice.

```python
bhk= [1, 2, 3, 4]
df_filtered = new_df[new_df['bhk'].isin(bhk)]
sns.kdeplot(data=df_filtered,x="price",hue="bhk",palette="Dark2",multiple="stack")
plt.xlim(0,600)
plt.show()
```

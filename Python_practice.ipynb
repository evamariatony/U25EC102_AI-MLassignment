Q6. Write a program to find the second largest number in a list.

```python
counts=int(input("enter the number of numbers to be appended"))
lst=[]
for i in range(counts):
    nums=int(input("enter numbers to list"))
    lst.append(nums)
print(lst)
lst.sort()
sec_largest_num=lst[-2]
print("the second largest number is:",sec_largest_num)


```

Q7. Write a program to remove duplicate elements from a list.

```python
lstt=[21,22,25,28,22,25,24,23]
new_list=[]
for i in lstt:
    if i not in new_list:
        new_list.append(i)
print(new_list)
```
